```python
import os
import sys
from pyspark.sql import SparkSession

os.environ["JAVA_HOME"] = "/Library/Java/JavaVirtualMachines/temurin-21.jdk/Contents/Home"
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

spark = SparkSession.builder.appName("RDD Transformation").getOrCreate()
print("Driver Python:", sys.executable)
print("Spark:", spark.version)
```

### WORD COUNT PROGRAM

```python
rdd = spark.sparkContext.textFile('/Users/darvikkunalbanda/DataEngineering/DE_Drill/pyspark/words.txt')

print(rdd.collect())
```

```python
new_rdd = rdd.flatMap(lambda x:x.split(' '))
print(new_rdd.collect())
```

```python
new_rdd1 = new_rdd.map(lambda x:(x,1))
print(new_rdd1.collect())
```

```python
new_rdd2 = new_rdd1.reduceByKey(lambda x,y : x+y)
print(new_rdd2.collect())
```

```python
## Single Line word count code

# flatmap()
# map()
# reduceByKey()
# collect()

new_rdd = print(rdd.flatMap(lambda x:x.split(' '))\
    .map(lambda x: (x,1))\
    .reduceByKey(lambda x,y : x+y)\
    .sortBy(lambda x:x[1] , ascending=False).collect())
```
