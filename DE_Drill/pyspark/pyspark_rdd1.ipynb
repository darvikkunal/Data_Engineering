```python
import os
import sys
from pyspark.sql import SparkSession

os.environ["JAVA_HOME"] = "/Library/Java/JavaVirtualMachines/temurin-21.jdk/Contents/Home"
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

spark = SparkSession.builder.appName("RDD Transformation").getOrCreate()
print("Driver Python:", sys.executable)
print("Spark:", spark.version)
```

RDD Transformations

```python
## Map Transformation
my_list = ['hello','jagan','sneha','shiva','meghana']

rdd = spark.sparkContext.parallelize(my_list)

print(rdd.collect())

new_rdd = rdd.map(lambda x :x.upper())

print(new_rdd.collect())
```

```python
my_numbers = [1,2,3,4,5]
rdd_num = spark.sparkContext.parallelize(my_numbers)
print(rdd_num.collect())
new_rdd_num = rdd_num.map(lambda x:x*10)
print(new_rdd_num.collect())
```

```python
# Filter Transformations
rdd = spark.sparkContext.parallelize([1,2,3,4,5,6])
print(rdd.collect())

filter_rdd = rdd.filter(lambda x:x>2)
print(filter_rdd.collect())
```

#FlatMap Transformations

```python
#python string to list

my_str='hello jagan good day!'
my_list=my_str.split(' ')
print(my_list)
```

```python
#FlatMap Transformations

rdd = spark.sparkContext.parallelize(['hello jagan','hello meghna','hello sneha'])
print(f'Actual RDD : {rdd.collect()} \n')
split_rdd = rdd.map(lambda x:x.split( )) # this will convert into list of list
print(f'List of list : {split_rdd.collect()} \n')
fm_rdd = rdd.flatMap(lambda x:x.split(' ')) # this will covert that list of list into a flatten list
print(f'Flatten list : {fm_rdd.collect()}')
```

```python
#GroupBy Key Transformations
my_list = [(25,5000), (30,70000), (25,6000), (35,90000), (30, 8000)]

rdd = spark.sparkContext.parallelize(my_list)
print(f' RDD : {rdd.collect()} \n')

grouped_rdd = rdd.groupByKey()
print(f' Grouped RDD : {grouped_rdd.collect()} \n')

grouped_rdd1 = grouped_rdd.mapValues(lambda x:sum(x))
print(f' Group by sum : {grouped_rdd1.collect()} \n ')
```

```python
#Reduce by Key is simillar to group by key
# In real time every one uses reduce by key - reduceByKey()

my_list = [(25,5000), (30,70000), (25,6000), (35,90000), (30, 8000)]

rdd = spark.sparkContext.parallelize(my_list)
print(f' RDD : {rdd.collect()} \n')

grouped_rdd = rdd.reduceByKey(lambda x,y: x+y)
print(f'Reduced by key : {grouped_rdd.collect()}')
```

```python
#Join Transformations
rdd1 = spark.sparkContext.parallelize([(1, ('Alice', 25)), (2, ('Bob', 30)), (3, ('Charlie', 35)),(4, ('Jagan', 50))])
rdd2 = spark.sparkContext.parallelize([(1, ('New York', 'Engineer')), (2, ('San Francisco', 'Artist')), (3, ('Boston','Developer'))])
joined_rdd = rdd1.join(rdd2)
print(joined_rdd.collect())
```

```python
#Distinct Transformations

rdd = spark.sparkContext.parallelize([1,2,3,4,3,4,3,2,1,4,5])
distinct_rdd = rdd.distinct()
print(distinct_rdd.collect())
```

```python
#Sortby Transformations

rdd = spark.sparkContext.parallelize([1,2,3,4,3,4,3,2,1,4,5,23,11,1,1,42,23,34,2354,36,457,456,35,6,45,4,52,3,5346,45,6,457,4,73,5,634,6,43,457,56,8,57,8,56,76,7,45,7])
sort_rdd = rdd.sortBy(lambda x:x, ascending=False) 
print(sort_rdd.collect())
```

```python
#Union Transformations

rdd1 = spark.sparkContext.parallelize([(1, ('Alice', 25)), (2, ('Bob', 30)), (3, ('Charlie', 35)),(4, ('Jagan', 50))])
rdd2 = spark.sparkContext.parallelize([(1, ('New York', 'Engineer')), (2, ('San Francisco', 'Artist')), (3, ('Boston','Developer'))])

union_rdd = rdd1.union(rdd2)
print(union_rdd.collect())
```

```python
# cartesian transformations

rdd1 = spark.sparkContext.parallelize([(1, ('Alice', 25)), (2, ('Bob', 30)), (3, ('Charlie', 35)),(4, ('Jagan', 50))])
rdd2 = spark.sparkContext.parallelize([(1, ('New York', 'Engineer')), (2, ('San Francisco', 'Artist')), (3, ('Boston','Developer'))])

cartesian_rdd = rdd1.cartesian(rdd2)
print(cartesian_rdd.collect())
```

```python
#Re-partition transformations

rdd = spark.sparkContext.parallelize([1,2,3,4,3,4,3,2,1,4,5])

print(rdd.getNumPartitions())

new_rdd = rdd.repartition(20)
print(new_rdd.getNumPartitions())
```

```python
#coalesce transformations

rdd = spark.sparkContext.parallelize([1,2,3,4,3,4,3,2,1,4,5])

new_rdd = rdd.coalesce(40)
print(new_rdd.getNumPartitions())

new_rdd1 = rdd.coalesce(12)
print(new_rdd1.getNumPartitions())
```
