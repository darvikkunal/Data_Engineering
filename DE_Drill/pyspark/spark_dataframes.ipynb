```python
import os
import sys
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

os.environ["JAVA_HOME"] = "/Library/Java/JavaVirtualMachines/temurin-21.jdk/Contents/Home"
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()
print("Driver Python:", sys.executable)
print("Spark:", spark.version)
```

## To create dataframes from a csv

```python
input_file_schema = StructType([
    StructField('product_id',StringType(),True),
    StructField('product_quantity',IntegerType(),True),
    StructField('product_price',IntegerType(),True)
])
```

```python
input_df = spark.read\
    .option('header',True)\
    .option('inferSchema',True)\
    .schema(input_file_schema)\
    .csv('/Users/darvikkunalbanda/DataEngineering/DE_Drill/dataset/orders.csv')
input_df.show()
```

```python
input_df.printSchema()
```

## To create dataframes from JSON

```python
json_df = spark.read.json('/Users/darvikkunalbanda/DataEngineering/DE_Drill/dataset/orders.json')
json_df.show()

json_df.printSchema()
```

```python
## Reading from nested JSON

nestedjson_df = spark.read.json('/Users/darvikkunalbanda/DataEngineering/DE_Drill/dataset/orders_nested.json')

nestedjson_df.printSchema()
```

```python
from pyspark.sql.functions import explode


flattenjson_df = nestedjson_df.select(
    'order_id','customer.*',explode('items').alias('item'),'status','total'
    ).select(
        'order_id','name','email','address.*','item.product','item.qty','status','total'
    )

flattenjson_df.show()
```
