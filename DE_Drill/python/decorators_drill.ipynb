```python
import time
from datetime import datetime, timedelta
```

```python
def timer_dec(base_fn):
    def enhanced_fn(*args, **kwargs):
        start_time = time.time()
        result = base_fn(*args,**kwargs)
        end_time = time.time()
        print(f"Task time: {end_time - start_time} seconds \n")
        return result
    return enhanced_fn

@timer_dec
def brew_tea(tea_type, sleep_time):
    print(f"Brewing {tea_type} tea...")
    time.sleep(sleep_time)
    print("Tea is ready!")

@timer_dec
def make_matcha():
    print("Making Matcha...")
    time.sleep(1)
    print("Matcha is ready!")
    return f"Drink matcha by {datetime.now() + timedelta(minutes=30)}"

brew_tea(tea_type="green", sleep_time=1)
print(make_matcha())
```

Problem 1
Create a decorator called mission_timer that:

Prints a 3 → 2 → 1 → Launch! countdown
Times how long the mission function takes
Works with two functions that take no arguments and return nothing
Then decorate and call:

launch_probe()
deploy_satellite()

```python
def mission_timer(base_fun):
    def enhanced_fn(*args,**kwargs):
        #countdown
        for i in range(3, 0 , -1):
            print(i)
            time.sleep(1)
        print("Launch \n")
        #mission time
        start_time = time.time()
        result = base_fun(*args,**kwargs)
        end_time = time.time()
        print(f"Task time : {end_time - start_time} seconds \n")
        return result
    return enhanced_fn


@mission_timer
def launch_probe(target):
    print(f"Launching probe toward {target}...!")
    time.sleep(1)
    return f"Probe successfully en route to {target}."

@mission_timer
def deploy_satellite(orbit_type,altitude_km):
    print(f"Deploying satellite into {orbit_type} orbit at {altitude_km} km...")
    time.sleep(2)

```

```python
launch_probe("Mars")
deploy_satellite("polar", 800)
deploy_satellite(orbit_type="geostationary", altitude_km=35786)
```

```python
deploy_satellite("polar", 800)

result = launch_probe("Europa")
print("Stored mission result:", result)
```

Final Decorator Exercise — retry

```python
def retry(base_fun):
    def wrapper(*args,**kwargs):
        for attempt in range(1,4):
            try:
                result = base_fun(*args,**kwargs)
                print(f"Success on attempt {attempt}")
                return result
            except Exception as e:
                print(f" Attempt : {attempt} \n Failed : {e}")
    return wrapper
```

```python
attempt_counter = 0

@retry
def fetch_data():
    global attempt_counter
    attempt_counter +=1

    if attempt_counter < 3:
        raise ConnectionError("Server not responding")
    return "Data is fetched successfully!"

fetch_data()
```

Print both results

```python
pythonsales = [
    {"product": "Laptop", "category": "Electronics", "amount": 85000},
    {"product": "Phone", "category": "Electronics", "amount": 45000},
    {"product": "Desk", "category": "Furniture", "amount": 12000},
    {"product": "Chair", "category": "Furniture", "amount": 8000},
    {"product": "Tablet", "category": "Electronics", "amount": 32000},
    {"product": "Lamp", "category": "Furniture", "amount": 3000},
]
```

```python
cat={}

for item in pythonsales:
    category = item['category']
    amount = item['amount']

    if category not in cat:
        cat[category] = amount
    else:
        cat[category] += amount

highest = max(pythonsales, key=lambda x:x['amount'])
print(f"Highest selling product: {highest['product']} - {highest['amount']}")


    

print(cat)
```

```python
# You have this messy list of employee records:
employees = [
    {"name": "Rahul", "dept": "Engineering", "salary": 95000},
    {"name": "Priya", "dept": "Marketing", "salary": 72000},
    {"name": "Amit", "dept": "Engineering", "salary": 110000},
    {"name": "Sara", "dept": "Marketing", "salary": 68000},
    {"name": "Kiran", "dept": "Engineering", "salary": 88000},
    {"name": "Neha", "dept": "HR", "salary": 61000},
]
```

```python
#Find average salary per department
dept_sal = {}

for i in employees:
    salary = i['salary']
    department = i['dept']
    name = i['name']

    if department not in dept_sal:
        dept_sal[department] = {
              'salaries': [],      # ← for average calculation
            'employees': []
        }

    dept_sal[department]['salaries'].append(salary)
    dept_sal[department]['employees'].append({'name':name, 'salary':salary})

    #print(dept_sal)

#Calculate salary
for dept, data in dept_sal.items():
    avg = sum(data['salaries']) / len(data['salaries'])

    highest = max(data['employees'], key = lambda x:x['salary'])
    print(f"{dept} : ")
    print(f"Average salary : {round(avg,2)}")
    print(f"Highest paid : {highest['name']} - {highest['salary']}")
```
