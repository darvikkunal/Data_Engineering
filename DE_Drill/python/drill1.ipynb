Problem 2 – Removing Duplicates

You have a list of website visitors, but some visitors appear more than once:

visitors = ["diana", "bob", "charlie", "diana", "bob", "alice", "charlie", "diana"]

Write code to get a list of unique visitors (no duplicates), sorted alphabetically. Print the result.

```python
visitors = ["diana", "bob", "charlie", "diana", "bob", "alice", "charlie", "diana"]
unique_visitors = list(set(visitors))
unique_visitors.sort(reverse=False)
unique_visitors
```

```python
def square(num):
    return num**2
```

```python
def transform_list(nums_list, transform_item):
    tansform_0 = transform_item(nums_list[0])
    tansform_1 = transform_item(nums_list[1])
    return [tansform_0, tansform_1]

mylist = [2,3]

transform_list(mylist, square)
```

Exercise 5: Sorting with Lambda
You are given a list of tuples, where each tuple contains a name and a score:

students = [('Alice', 88), ('Bob', 72), ('Charlie', 95)]
Use the sorted() function and a lambda to sort the list by score in descending order.

```python
students = [('Alice', 88), ('Bob', 72), ('Charlie', 95)]

sorted_students =sorted(students , key=lambda x : x[1], reverse=True)
print(sorted_students)
```

Problem 3

You want to look back at your Python-related entries.

Using context managers (with statements), read journal.txt and create a new file called python_entries.txt containing only the entries that mention "Python". 

Then read the new file back and print its contents.

```python
# Step 1 : Read original file and filter lines
with open('/Users/darvikkunalbanda/DataEngineering/DE_Drill/dataset/entries.txt','r') as infile:
    lines = infile.readlines()

# Step 2: Keep only lines containing "Python"
python_line = [line for line in lines if "Python" in line]

#Step 3: Write filtered lines to new file
with open('/Users/darvikkunalbanda/DataEngineering/DE_Drill/dataset/python_entries.txt','w') as outfile:
    outfile.writelines(python_line)

#Step 4 : Read and print new files
with open('/Users/darvikkunalbanda/DataEngineering/DE_Drill/dataset/python_entries.txt','r') as file:
    contents = file.read()

print(contents)
```
